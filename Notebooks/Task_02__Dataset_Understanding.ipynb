# Part 6 – Data Quality Assessment

##  Check missing values

```python
missing_values = df.isnull().sum()

missing_summary = pd.DataFrame({
    'Missing Values': missing_values,
    'Missing Percentage': (
        missing_values / len(df) * 100
    ).round(2)
})

missing_summary = missing_summary[
    missing_summary['Missing Values'] > 0
]

display(missing_summary)
```

## Visualize missing values

```python
plt.figure(figsize=(12, 5))

sns.heatmap(
    df.isnull(),
    cbar=False,
    cmap='viridis',
    yticklabels=False
)

plt.title('Missing Values in SmartCare Hospital Dataset')
plt.xlabel('Dataset Attributes')
plt.ylabel('Patient Records')

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

##  Check duplicate records

```python
duplicate_count = df.duplicated().sum()

print("Total Duplicate Records:", duplicate_count)

if duplicate_count == 0:
    print("No duplicate records found.")
else:
    print("Duplicate records detected.")
```

## Check duplicate patient IDs

```python
duplicate_patients = df['patient_id'].duplicated().sum()

print("Duplicate Patient IDs:", duplicate_patients)

print(
    "Unique Patients:",
    df['patient_id'].nunique()
)
```

##  Check data types

```python
data_types = pd.DataFrame({
    'Attribute': df.columns,
    'Data Type': df.dtypes.values
})

display(data_types)
```

## Check unique categorical values

```python
categorical_columns = df.select_dtypes(
    include=['object']
).columns

for column in categorical_columns:

    print("\nAttribute:", column)

    print("Unique Values:",
          df[column].nunique())

    print(df[column].value_counts(dropna=False))
```

##  Check target variable distribution

```python
target_counts = df['no_show'].value_counts()

target_percentage = (
    df['no_show'].value_counts(normalize=True) * 100
)

target_summary = pd.DataFrame({
    'Count': target_counts,
    'Percentage': target_percentage.round(2)
})

display(target_summary)
```

## Visualize target distribution

```python
plt.figure(figsize=(7, 5))

ax = sns.countplot(
    data=df,
    x='no_show'
)

plt.title('Appointment No-Show Target Distribution')
plt.xlabel('No-Show (0 = Attended/Other, 1 = No-Show)')
plt.ylabel('Number of Records')

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()
plt.show()
```

##  Check appointment status and target consistency

```python
status_target = pd.crosstab(
    df['appointment_status'],
    df['no_show'],
    margins=True
)

display(status_target)
```

## Detect potential numerical outliers

```python
numerical_columns = df.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

# Exclude identifiers and prediction targets
exclude = [
    'record_id',
    'no_show',
    'readmitted_30_days'
]

numerical_features = [
    col for col in numerical_columns
    if col not in exclude
]

outlier_results = []

for column in numerical_features:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    outlier_results.append({
        'Attribute': column,
        'Lower Limit': round(lower_limit, 2),
        'Upper Limit': round(upper_limit, 2),
        'Potential Outliers': len(outliers)
    })

outlier_df = pd.DataFrame(outlier_results)

display(outlier_df)
```

## Check appointment dates

```python
df['appointment_date'] = pd.to_datetime(
    df['appointment_date'],
    errors='coerce'
)

print("Earliest Appointment:",
      df['appointment_date'].min())

print("Latest Appointment:",
      df['appointment_date'].max())

print("Invalid Dates:",
      df['appointment_date'].isnull().sum())
```

##  Generate a data quality summary

```python
quality_summary = pd.DataFrame({
    'Quality Check': [
        'Total Records',
        'Total Attributes',
        'Missing Values',
        'Duplicate Rows',
        'Unique Patients',
        'Missing Target Values',
        'Invalid Appointment Dates'
    ],

    'Results': [
        len(df),
        len(df.columns),
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        df['patient_id'].nunique(),
        df['no_show'].isnull().sum(),
        df['appointment_date'].isnull().sum()
    ]
})

display(quality_summary)
```
